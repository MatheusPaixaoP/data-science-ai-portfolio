## 2. Text Representation & Pattern Construction
Text samples are converted into feature vectors capturing language-specific morphological, orthographic, and syntactic properties.

```python
import random

pre_padroes = []
for frase in ingles:
  pre_padroes.append( [frase, 'inglês'])

for frase in espanhol:
  pre_padroes.append( [frase, 'espanhol'])

for frase in portugues:
  pre_padroes.append( [frase, 'português'])

random.shuffle(pre_padroes)
print(pre_padroes)
```

## 4. Feature Extraction Pipeline
Engineering 17 distinct linguistic features using Regular Expressions (Regex):
- Average word length
- Spanish inverted punctuation (, )
- Language-specific character indicators (, )
- Language suffixes ( vs. )
- English contractions ()
- Stopword frequency distributions
- Accent mark counters (, , , , , , )

```python
# a entrada é o vetor pre_padroes e a saída desse passo deverá ser "padrões"
import re

def tamanhoMedioFrases(texto):
  palavras = re.split("\s",texto)
  #print(palavras)
  tamanhos = [len(s) for s in palavras if len(s)>0]
  #print(tamanhos)
  soma = 0
  for t in tamanhos:
    soma=soma+t
  return soma / len(tamanhos)

def has_spanish_inverted_punctuation(texto):
    return 1.0 if re.search(r"[¿¡]", texto, re.UNICODE) else 0.0

def has_cedilla(texto):
    return 1.0 if 'ç' in texto else 0.0

def has_enie(texto):
    return 1.0 if 'ñ' in texto else 0.0

def has_portuguese_suffix(texto):
    return 1.0 if texto.lower().endswith('ção') else 0.0

def has_spanish_suffix(texto):
    return 1.0 if texto.lower().endswith('ción') else 0.0

def has_english_contraction(texto):
    import re
    return 1.0 if re.search(r"\b\w+'\w+\b", texto) else 0.0

def contains_common_english(texto):
    import re
    return 1.0 if re.search(r"\b(the|how|is|and)\b", texto, re.IGNORECASE) else 0.0

def contains_common_spanish(texto):
    import re
    return 1.0 if re.search(r"\b(el|la|que|de|y)\b", texto, re.IGNORECASE) else 0.0

def contains_common_portuguese(texto):
    import re
    return 1.0 if re.search(r"\b(que|não|por|para)\b", texto, re.IGNORECASE) else 0.0

def count_portuguese_accents(texto):
    return sum(texto.count(c) for c in ['ã', 'õ', 'â', 'ê', 'ô', 'á', 'é', 'í', 'ó', 'ú', 'ç'])

def count_spanish_accents(texto):
    return sum(texto.count(c) for c in ['á', 'é', 'í', 'ó', 'ú', 'ñ'])

def count_tildes(texto):
    return texto.count('ñ') + texto.count('ã') + texto.count('õ')

def contains_spanish_articles(texto):
    import re
    return 1.0 if re.search(r'\b(el|la|los|las|un|una|unos|unas)\b', texto, re.IGNORECASE) else 0.0

def contains_portuguese_articles(texto):
    import re
    return 1.0 if re.search(r'\b(o|a|os|as|um|uma|uns|umas)\b', texto, re.IGNORECASE) else 0.0

def contains_spanish_question_words(texto):
    return 1.0 if any(p in texto.lower() for p in ['¿qué', '¿dónde', '¿cuál', '¿cómo']) else 0.0

def contains_portuguese_question_words(texto):
    return 1.0 if any(p in texto.lower() for p in ['o que', 'onde', 'qual', 'como']) else 0.0

def y_ratio(texto):
    import re
    words = re.findall(r"\b\w+\b", texto.lower())
    if not words: return 0.0
    return sum(1 for w in words if 'y' in w) / len(words)

def extraiCaracteristicas(frase):
  # frase é um vetor [ 'texto', 'lingua' ]
  texto = frase[0]
  pattern_regex = re.compile('[^\w+]', re.UNICODE)
  texto = re.sub(pattern_regex,' ',texto)
  #print(texto)
  caracteristica1=tamanhoMedioFrases(texto)
  caracteristica2=has_spanish_inverted_punctuation(texto)
  caracteristica3=has_cedilla(texto)
  caracteristica4=has_enie(texto)
  caracteristica5=has_portuguese_suffix(texto)
  caracteristica6=has_spanish_suffix(texto)
  caracteristica7=has_english_contraction(texto)
  caracteristica8=contains_common_english(texto)
  caracteristica9=contains_common_spanish(texto)
  caracteristica10=contains_common_portuguese(texto)
  caracteristica11=count_portuguese_accents(texto)
  caracteristica12=count_spanish_accents(texto)
  caracteristica13=count_tildes(texto)
  caracteristica14=contains_spanish_articles(texto)
  caracteristica15=contains_spanish_question_words(texto)
  caracteristica16=contains_portuguese_question_words(texto)
  caracteristica17=contains_portuguese_articles(texto)

  # acrescente as suas funcoes no vetor padrao
  padrao = [valor for nome, valor in locals().items() if nome.startswith("caracteristica")]
  padrao.append(frase[1])  # adiciona o rótulo ao final

  return padrao

def geraPadroes(frases):
  padroes = []
  for frase in frases:
    padrao = extraiCaracteristicas(frase)
    padroes.append(padrao)
  return padroes

# converte o formato [frase classe] em
# [caracteristica_1, caracteristica_2,... caracteristica n, classe]
padroes = geraPadroes(pre_padroes)

#
# apenas para visualizacao
print(padroes)

dados = pd.DataFrame(padroes)
dados
```

## 5. Model Training & Train-Test Partitioning
Splitting dataset into 75% Training and 25% Test sets with stratified sampling.

```python
from sklearn.model_selection import train_test_split
import numpy as np


#from sklearn.metrics import confusion_matrix

vet = np.array(padroes)
classes = vet[:,-1]         # classes = [p[-1] for p in padroes]
#print(classes)
padroes_sem_classe = vet[:,0:-1]
#print(padroes_sem_classe)
X_train, X_test, y_train, y_test = train_test_split(padroes_sem_classe, classes, test_size=0.25, stratify=classes)
```

## 6. Model Evaluation (Support Vector Machine)
Evaluating SVM classifier performance using Confusion Matrix and Classification Metrics.

```python
from sklearn import svm
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

treinador = svm.SVC()  #algoritmo escolhido
modelo = treinador.fit(X_train, y_train)

#
# score com os dados de treinamento
acuracia = modelo.score(X_train, y_train)
print("Acurácia nos dados de treinamento: {:.2f}%".format(acuracia * 100))

#
# melhor avaliar com a matriz de confusão
y_pred = modelo.predict(X_train)
cm = confusion_matrix(y_train, y_pred)
print(cm)
print(classification_report(y_train, y_pred))

#
# com dados de teste que não foram usados no treinamento
print('métricas mais confiáveis')
y_pred2 = modelo.predict(X_test)
cm = confusion_matrix(y_test, y_pred2)
print(cm)
print(classification_report(y_test, y_pred2))
```
